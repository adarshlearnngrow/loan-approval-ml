```python
from imports import *
import json

query = "SELECT COLUMN_NAME FROM INFORMATION_SCHEMA.COLUMNS \
where TABLE_NAME = 'preprocessed_data' and DATA_TYPE = 'varchar';"
cat_cols = sql_connect(query)

cat_col_mapping = {}
for col in cat_cols['COLUMN_NAME']:
    query = f"SELECT DISTINCT {col} FROM preprocessed_data;"
    values = sql_connect(query)
    cat_col_mapping[col] = values[col].tolist()

categorical_cols = list(cat_col_mapping.keys())
categories = [cat_col_mapping[col] for col in categorical_cols]

with open('categories.json', 'w') as f:
    json.dump(cat_col_mapping, f, indent=2)

print("Saved categories for all categorical columns to categories.json")   
```
