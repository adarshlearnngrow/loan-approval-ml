```python
from imports import *
from preprocessing_util import get_column_groups, get_transformers
from sklearn.model_selection import StratifiedKFold, cross_validate, cross_val_predict, RandomizedSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline as SkPipeline
from sklearn.compose import ColumnTransformer
from sklearn.metrics import (
    precision_recall_curve, f1_score, average_precision_score,
    classification_report, confusion_matrix, precision_score, recall_score
)
from xgboost import XGBClassifier
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler
from sklearn.ensemble import RandomForestClassifier
import mlflow
import mlflow.sklearn
```

```python
query = "select * from train_data"
train_data = sql_connect(query)
X = train_data.drop(columns=['TARGET'])
y = train_data['TARGET']
del train_data
X.head()
```

```python
query = "select * from test_data"
test_data = sql_connect(query)
X_test = test_data.drop(columns=['TARGET'])
y_test = test_data['TARGET']
del test_data
X.head()
```

```python
cv_strategy = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
```

```python
column_groups = get_column_groups(X)
```

#### Model Definitions

```python
model_def = [
    {
        'name': 'Logistic Regression L1',
        'model': LogisticRegression(max_iter=1000, penalty='l1', solver='liblinear', class_weight='balanced'),
        'param_grid': {'model__C': [0.01, 0.1, 1]},
        'sampler': None
    },
    {
        'name': 'Logistic Regression L1 with SMOTE',
        'model': LogisticRegression(max_iter=1000, penalty='l1', solver='liblinear', class_weight='balanced'),
        'param_grid': {'model__C': [0.01, 0.1, 1]},
        'sampler': SMOTE(sampling_strategy='minority', random_state=42)
    },
    {
        'name': 'Logistic Regression L1 with UnderSampling',
        'model': LogisticRegression(max_iter=1000, penalty='l1', solver='liblinear', class_weight='balanced'),
        'param_grid': {'model__C': [0.01, 0.1, 1]},
        'sampler': RandomUnderSampler(sampling_strategy='majority', random_state=42)
    },
    {
        'name': 'Random Forest',
        'model': RandomForestClassifier(class_weight='balanced', random_state=42),
        'param_grid': {
            'model__n_estimators': [50, 100, 200],
            'model__max_depth': [None, 10, 20],
            'model__min_samples_split': [2, 5, 10],
            'model__min_samples_leaf': [1, 2, 4]
        },
        'sampler': None
    },
    {
        'name': 'Random Forest with SMOTE',
        'model': RandomForestClassifier(class_weight='balanced', random_state=42),
        'param_grid': {
            'model__n_estimators': [50, 100, 200],
            'model__max_depth': [None, 10, 20],
            'model__min_samples_split': [2, 5, 10],
            'model__min_samples_leaf': [1, 2, 4]
        },
        'sampler': SMOTE(sampling_strategy='minority', random_state=42)
    },
    {
        'name': 'Random Forest with UnderSampling',
        'model': RandomForestClassifier(class_weight='balanced', random_state=42),
        'param_grid': {
            'model__n_estimators': [50, 100, 200],
            'model__max_depth': [None, 10, 20],
            'model__min_samples_split': [2, 5, 10],
            'model__min_samples_leaf': [1, 2, 4]
        },
        'sampler': RandomUnderSampler(sampling_strategy='majority', random_state=42)
    },
    {
        'name': 'XGBoost',
        'model': XGBClassifier(use_label_encoder=False, eval_metric='logloss'),
        'param_grid': {
            'model__n_estimators': [50, 100, 200],
            'model__max_depth': [3, 5, 7],
            'model__learning_rate': [0.01, 0.1, 0.2],
            'model__subsample': [0.8, 1.0]
        },
        'sampler': None
    },
    {
        'name': 'XGBoost with SMOTE',
        'model': XGBClassifier(use_label_encoder=False, eval_metric='logloss'),
        'param_grid': {
            'model__n_estimators': [50, 100, 200],
            'model__max_depth': [3, 5, 7],
            'model__learning_rate': [0.01, 0.1, 0.2],
            'model__subsample': [0.8, 1.0]
        },
        'sampler': SMOTE(sampling_strategy='minority', random_state=42)
    },
    {
        'name': 'XGBoost with UnderSampling',
        'model': XGBClassifier(use_label_encoder=False, eval_metric='logloss'),
        'param_grid': {
            'model__n_estimators': [50, 100, 200],
            'model__max_depth': [3, 5, 7],
            'model__learning_rate': [0.01, 0.1, 0.2],
            'model__subsample': [0.8, 1.0]
        },
        'sampler': RandomUnderSampler(sampling_strategy='majority', random_state=42)
    }
]
```

```python
mlflow.set_experiment("Credit Risk Modelling Project")
mlflow.set_tracking_uri("http://127.0.0.1:5000/") 

for i in range(len(model_def)):
    print(f"Training model : {model_def[i]['name']}")
    if "Regression" in model_def[i]['name']:
        transformers = get_transformers(
            column_groups['log_col'],
            column_groups['scale_only_cols'],
            column_groups['categorical_cols'],
            column_groups['binary_cols'],
            'linear'
        )

    else:
        transformers = get_transformers(
            column_groups['log_col'],
            column_groups['scale_only_cols'],
            column_groups['categorical_cols'],
            column_groups['binary_cols'],
            'tree'
        )

    if model_def[i]['sampler'] is not None:
        pipeline = ImbPipeline([
            ('transformer', ColumnTransformer(transformers)),
            ('sampler', model_def[i]['sampler']),
            ('model', model_def[i]['model'])
        ])
    else:
        pipeline = SkPipeline([
            ('transformer', ColumnTransformer(transformers)),
            ('model', model_def[i]['model'])
        ])

    random_search = RandomizedSearchCV(
        estimator=pipeline,
        param_distributions=model_def[i]['param_grid'],
        n_iter=5,  
        scoring="average_precision",
        cv=cv_strategy,
        n_jobs=-1,
        random_state=42
    )
    
    with mlflow.start_run(run_name=model_def[i]['name']) as run:
        
        random_search.fit(X, y)
        best_model = random_search.best_estimator_
        
        y_prob = best_model.predict_proba(X)[:, 1]
        
        precision, recall, thresholds = precision_recall_curve(y, y_prob)
        ap_score  = average_precision_score(y, y_prob)

        f1_curve = 2 * (precision * recall) / (precision + recall + 1e-10)
        f1_score_max = np.max(f1_curve)
        f1_threshold = thresholds[np.argmax(f1_curve)]

        y_pred = (y_prob >= f1_threshold).astype(int)

        report = classification_report(y, y_pred, output_dict=True)
        report_df = pd.DataFrame(report).transpose()
        report_df.to_csv("classification_report.csv")

        cm = confusion_matrix(y, y_pred)
        cm_df = pd.DataFrame(cm)
        cm_df.to_csv("confusion_matrix.csv")

        preds_df = pd.DataFrame({'y_true': y, 'y_pred': y_pred, 'y_proba': y_prob})
        preds_df.to_csv("predictions.csv", index=False)

        # Log model hyperparameters and metrics
        mlflow.log_params(random_search.best_params_)
        mlflow.log_metric("average_precision_score", ap_score)
        mlflow.log_metric("f1_score_max", f1_score_max)
        mlflow.log_metric("f1_threshold", f1_threshold)

        # Log artifacts
        mlflow.log_artifact("classification_report.csv")
        mlflow.log_artifact("confusion_matrix.csv") 
        mlflow.log_artifact("predictions.csv")

        # Log the model
        mlflow.sklearn.log_model(best_model, "model")

        print(f"Model {model_def[i]['name']} trained and logged successfully.")

        # ===== LOG TEST METRICS TO MLFLOW =====

        y_test_prob = best_model.predict_proba(X_test)[:, 1]
        y_test_pred = (y_test_prob >= f1_threshold).astype(int) 

        test_precision = precision_score(y_test, y_test_pred)
        test_recall = recall_score(y_test, y_test_pred)
        test_f1 = f1_score(y_test, y_test_pred)
        test_avg_precision = average_precision_score(y_test, y_test_prob)

        mlflow.log_metric("test_precision", test_precision)
        mlflow.log_metric("test_recall", test_recall)
        mlflow.log_metric("test_f1", test_f1)
        mlflow.log_metric("test_avg_precision", test_avg_precision)

        report_test = classification_report(y_test, y_test_pred, output_dict=True)
        pd.DataFrame(report_test).transpose().to_csv("classification_report_test.csv")
        mlflow.log_artifact("classification_report_test.csv")

        cm_test = confusion_matrix(y_test, y_test_pred)
        pd.DataFrame(cm_test).to_csv("confusion_matrix_test.csv")
        mlflow.log_artifact("confusion_matrix_test.csv")

        preds_test_df = pd.DataFrame({'y_true': y_test, 'y_pred': y_test_pred, 'y_proba': y_test_prob})
        preds_test_df.to_csv("predictions_test.csv", index=False)
        mlflow.log_artifact("predictions_test.csv")

        print("Test metrics and artifacts logged.")
```
